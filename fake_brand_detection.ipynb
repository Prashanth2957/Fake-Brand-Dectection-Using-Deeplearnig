```python
import os
import cv2

def convert_images(folder):
    for root, dirs, files in os.walk(folder):
        for file in files:
            path = os.path.join(root, file)

            if file.lower().endswith((".png",".jpeg",".jpg",".webp")):
                img = cv2.imread(path)

                new_path = os.path.splitext(path)[0] + ".jpg"
                cv2.imwrite(new_path, img)

                if not path.endswith(".jpg"):
                    os.remove(path)

convert_images("dataset")

print("All images converted to JPG")
```

```python
import os
import cv2

dataset_path = "/content/dataset/train"

for root, dirs, files in os.walk(dataset_path):
    for file in files:
        path = os.path.join(root, file)
        try:
            img = cv2.imread(path)
            if img is None or img.shape[0] < 10 or img.shape[1] < 10:
                os.remove(path)
                print("Removed corrupt image:", path)
        except:
            os.remove(path)
            print("Removed broken image:", path)

print("Dataset cleaned successfully")
```

```python
import os
import shutil
import random

train_dir = "/content/dataset/train"
val_dir = "/content/dataset/val"

os.makedirs(val_dir, exist_ok=True)

for class_name in os.listdir(train_dir):
    class_train = os.path.join(train_dir, class_name)
    class_val = os.path.join(val_dir, class_name)

    os.makedirs(class_val, exist_ok=True)

    images = os.listdir(class_train)
    random.shuffle(images)

    val_count = int(len(images) * 0.2)

    for img in images[:val_count]:
        shutil.move(os.path.join(class_train, img),
                    os.path.join(class_val, img))

print("Validation dataset created")
```

```python
from ultralytics import YOLO

model = YOLO("yolov8s-cls.pt")

model.train(
    data="dataset",
    epochs=50,
    imgsz=320,
    batch=16,
    val=False
)
```
